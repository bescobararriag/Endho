```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# GASTO PROMEDIO POR DIA PRESA ENDHO (m3/s) CSV
df = pd.read_csv("HISTORICO_PRESA_ENDHO.csv")

# borrar espacios en blanco entre tablas
df = df.dropna(subset=["GASTO PROMEDIO POR DIA PRESA ENDHO"])

# borrar fila de unidades (m3/S)
df = df.drop([0], axis=0).reset_index(drop=True)
```

```python
# crear dataframes separados para cada año (2016-2026)
rows_per_year = 33
df_years = []

for i in range(11):
    start = i*rows_per_year
    end = start + rows_per_year
    df_year = df.iloc[start:end].copy()
    df_years.append(df_year)

    # create dictionary 
    years = {
        2016 + i: df_years[i]
        for i in range(len(df_years))
    }
```

```python
months = ["ENERO", "FEBRERO", "MARZO", "ABRIL", "MAYO", "JUNIO", "JULIO", "AGOSTO", "SEPTIEMBRE", "OCTUBRE", "NOVIEMBRE", "DICIEMBRE"]
flows = ["DERRAMES", "DESFOGUES", "CONCECION", "DISTRITO 100", "DISTRITO 003"]
years = [2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025, 2026]

# lista para todos los años
dfs_all_years = []

for j, year in enumerate(years):

    # lista para todos los meses en un año
    dfs_one_year = []

    for i, month in enumerate(months):
        # enero inicia en columna 0, febrero en columna 6
        start = 0 if i == 0 else 6 + (i-1) *5
        end = start + (6 if i == 0 else 5)

        # fecha mensual 
        df_month = df_years[j].iloc[:, start:end].copy()

        # quitar primera fila sin datos
        df_month = df_month.iloc[2:].copy()

        # renombrar columnas
        if i == 0:
            df_month.columns = ["DIA"] + flows
        else:
            df_month.columns = flows
            df_month.insert(0, "DIA", range(1, len(df_month)+1))

        df_month.insert(1, "MES", month)
        df_month.insert(2, "AÑO", year)
        dfs_one_year.append(df_month)

    df_one_year_long = pd.concat(dfs_one_year, ignore_index=True)
    dfs_all_years.append(df_one_year_long)

df_all_years_long = pd.concat(dfs_all_years, ignore_index=True)
```

```python
# typos 

df_all_years_long = df_all_years_long.replace("8..525", 8.525)
df_all_years_long = df_all_years_long.replace(".6.725", 6.725)
```

```python
# export as csv
df_all_years_long.to_csv("HISTORICO_PRESA_ENDHO_longform.csv", encoding="utf-8")
```
